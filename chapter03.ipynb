## 리스트와 튜플

### 리스트

- 대괄호 [ ]  *추가, 수정 가능

### 튜플 

- 소괄호 ( )   * 수정 불가 (보통 고정값을 사용)

### 딕셔너리 

- 중괄호 { }   

````python
#리스트 : 가변 데이터 저장 
shopping_list = ["apple", "banana", "cherry"]

print(shopping_list)
print(shopping_list[0])

shopping_list[1] = "kiwi"
print(shopping_list[0])
print(shopping_list)
````

---
### 리스트 메서드

````python
# append
fruits = ["apple"]
fruits.append("bananan")
print(fruits)

#insert
fruits.insert(0, "kiwi")
print(fruits)

#remove
fruits.remove("apple")
print(fruits)

# fruits.remove("grape")
````

### 튜플 (Tuple)
- 한 번 생성하면 요소를 추가, 수정, 삭제 할 수 없다. 

````python
axis = (38.123, 41.123)

# axis[0] = 36.1234
# 위 문구를 실행하면 "오류" : 튜플을 직접 수정할 수 없음 

point = 1, 2, 3
print(point, type(point))
````

````python
import sys
print(sys.getsizeof([1, 2, 3]))
print(sys.getsizeof((1, 2, 3)))  # 튜플 : 데이터 안정성 
````

### 슬라이싱

````python
numbers = [0, 1, 2, 3, 4, 5]

# 1번부터 4번 미만까지 
numbers[1:4]

# 2칸 간격으로 추출
numbers[::2]

# 리스트 역순으로 추출
numbers[::-1]

# 앞에서 3개만 추출
numbers[:3]

# 3번부터 끝까지 추출
numbers[3:]
````

````python
print((10, 20, 30, 40)[1:3])
````
